```python
from PIL import Image, ImageDraw, ImageFont

def generate_base_template():
    # Create a blank white landscape image (A4 ratio approx 1600x1200)
    width, height = 1600, 1200
    img = Image.new('RGB', (width, height), color='#FFFFFF')
    draw = ImageDraw.Draw(img)

    # Add a formal double border
    draw.rectangle([30, 30, width - 30, height - 30], outline='#003366', width=8)
    draw.rectangle([45, 45, width - 45, height - 45], outline='#D4AF37', width=4)

    # Save as your base template
    img.save("certificate_template.jpg")
    print("Base certificate_template.jpg generated successfully!")

generate_base_template()
```
